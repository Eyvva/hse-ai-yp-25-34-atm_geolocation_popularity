```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('/Users/anastasia/Downloads/train_with_new_features.csv')
df = df.drop(columns=['id', 'address', 'address_rus'])
df = df.dropna()
df
```

```python
X_train = df.drop(columns=['target'])
y_train = df['target']
```

```python
df_test = pd.read_csv('/Users/anastasia/Downloads/test_with_new_features.csv')
df_test = df_test.drop(columns=['Unnamed: 0', 'id', 'address', 'address_rus'])
df_test = df_test.dropna()
df_test
```

```python
X_test = df_test.drop(columns=['target'])
Y_test = df_test['target']
```

```python
column_names = X_train.columns
column_names
```

```python
X_train['population'] = X_train['population'].str.replace("\xa0", "").astype(float)
```

## LinearRegression

```python
model = LinearRegression()
model.fit(X_train.astype(float), y_train)
y_pred = model.predict(X_test[column_names])
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error, mean_absolute_error
from sklearn.metrics import r2_score
import numpy as np

mse_model = mean_squared_error(Y_test, y_pred)
mape_model = mean_absolute_percentage_error(Y_test, y_pred)
mae_model = mean_absolute_error(Y_test, y_pred)
r2_model = r2_score(Y_test, y_pred)
print(f"MSE: {mse_model:.4f}")
print(f"MAPE: {mape_model:.4f}")
print(f"MAE: {mae_model:.4f}")
print(f"R² (коэффициент детерминации): {r2_model:.6f}")
print(f"R² в процентах: {r2_model * 100:.2f}%")
```

##### Средняя квадратичная ошибка составляет 0.0030 квадратных единиц. Средняя абсолютная процентная ошибка составляет 12.34%. Средняя абсолютная ошибка составляет 0.0485 единиц. Модель объясняет только 37.43% дисперсии целевой переменной

## SVM

```python
from sklearn import svm
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test[column_names])
model = svm.SVR(kernel='rbf', C=1.0, gamma='scale')
model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test[column_names])
```

```python
mse_svm = mean_squared_error(Y_test, y_pred)
mape_svm = mean_absolute_percentage_error(Y_test, y_pred)
mae_svm = mean_absolute_error(Y_test, y_pred)
r2_svm = r2_score(Y_test, y_pred)
print(f"MSE: {mse_svm:.4f}")
print(f"MAPE: {mape_svm:.4f}")
print(f"MAE: {mae_svm:.4f}")
print(f"R² (коэффициент детерминации): {r2_svm:.6f}")
print(f"R² в процентах: {r2_svm * 100:.2f}%")
```

##### Средняя квадратичная ошибка составляет 0.0073 квадратных единиц. Средняя абсолютная процентная ошибка составляет 21.0175%. Средняя абсолютная ошибка составляет 0.0812 единиц. Модель объясняет только -50.02% дисперсии целевой переменной.

##### То есть модель дает результат сильно хуже, чем простое предсказание средним значением

## DecisionTreeRegressor

```python
from sklearn.tree import DecisionTreeRegressor
dt_reg = DecisionTreeRegressor(
    max_depth=5,            # Ограничение глубины
    min_samples_split=10,   # Минимум samples для разделения
    min_samples_leaf=5,     # Минимум samples в листе
    random_state=42
)
dt_reg.fit(X_train, y_train)
y_pred = dt_reg.predict(X_test[column_names])
```

```python
mse_dt_reg = mean_squared_error(Y_test, y_pred)
mape_dt_reg = mean_absolute_percentage_error(Y_test, y_pred)
mae_dt_reg = mean_absolute_error(Y_test, y_pred)
r2_dt_reg = r2_score(Y_test, y_pred)
print(f"MSE: {mse_dt_reg:.4f}")
print(f"MAPE: {mape_dt_reg:.4f}")
print(f"MAE: {mae_dt_reg:.4f}")
print(f"R² (коэффициент детерминации): {r2_dt_reg:.6f}")
print(f"R² в процентах: {r2_dt_reg * 100:.2f}%")
```

##### Использование рещающего дерева сильно улучшило результат: MSE, MAPE и MAE значительно упали, а доля объясненной дисперсии увеличилась до 92.79

## RandomForestRegressor

```python
from sklearn.ensemble import RandomForestRegressor
regr = RandomForestRegressor(max_depth=5, random_state=42)
regr.fit(X_train, y_train)
y_pred = regr.predict(X_test[column_names])
```

```python
mse_regr = mean_squared_error(Y_test, y_pred)
mape_regr = mean_absolute_percentage_error(Y_test, y_pred)
mae_regr = mean_absolute_error(Y_test, y_pred)
r2_regr = r2_score(Y_test, y_pred)
print(f"MSE: {mse_regr:.4f}")
print(f"MAPE: {mape_regr:.4f}")
print(f"MAE: {mae_regr:.4f}")
print(f"R² (коэффициент детерминации): {r2_regr:.6f}")
print(f"R² в процентах: {r2_regr * 100:.2f}%")
```

## GradientBoostingRegressor

```python
from sklearn.ensemble import GradientBoostingRegressor
gbr = GradientBoostingRegressor(
    n_estimators=100,       # Количество деревьев
    learning_rate=0.1,      # Скорость обучения
    max_depth=3,            # Глубина каждого дерева
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42
)
gbr.fit(X_train, y_train)
y_pred = gbr.predict(X_test[column_names])
```

```python
mse_gbr = mean_squared_error(Y_test, y_pred)
mape_gbr = mean_absolute_percentage_error(Y_test, y_pred)
mae_gbr = mean_absolute_error(Y_test, y_pred)
r2_gbr = r2_score(Y_test, y_pred)
print(f"MSE: {mse_gbr:.4f}")
print(f"MAPE: {mape_gbr:.4f}")
print(f"MAE: {mae_gbr:.4f}")
print(f"R² (коэффициент детерминации): {r2_gbr:.6f}")
print(f"R² в процентах: {r2_gbr * 100:.2f}%")
```

## GridSearchCV

Попробуем подобрать лучшие гиперпараметры для KNN

```python
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsRegressor
```

```python
parameters = {
    'n_neighbors': [3, 5, 7, 9, 11],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan']
}

base_reg = KNeighborsRegressor()
reg_cv = GridSearchCV(base_reg, parameters)
reg_cv.fit(X_train, y_train)
```

```python
reg_cv.best_estimator_
```

```python
y_pred = reg_cv.predict(X_test[column_names])
```

```python
mse_knn = mean_squared_error(Y_test, y_pred)
mape_knn = mean_absolute_percentage_error(Y_test, y_pred)
mae_knn = mean_absolute_error(Y_test, y_pred)
r2_knn = r2_score(Y_test, y_pred)
print(f"MSE: {mse_knn:.4f}")
print(f"MAPE: {mape_knn:.4f}")
print(f"MAE: {mae_knn:.4f}")
print(f"R² (коэффициент детерминации): {r2_knn:.6f}")
print(f"R² в процентах: {r2_knn * 100:.2f}%")
```
